## Image Enhancement of Low-Contrast Image

This project implements Histogram Equalization to enhance the contrast of a low-contrast image.

The original image is analyzed using its grayscale representation and histogram. Histogram Equalization is then applied to spread the pixel intensity values and improve the contrast of the image.

The original and enhanced images are compared visually and through their histograms.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
```

## 1. Mount Google Drive

The input image is stored in Google Drive inside the `PCD_Assignment02` folder. Google Drive is mounted so that the image can be accessed directly from the notebook.

```python
drive.mount('/content/drive')
```

## 2. Load the Original Image

The `city.jpg` image is loaded from the PCD_Assignment02 folder in Google Drive.

```python
image_path = '/content/drive/MyDrive/PCD_Assignment02/city.jpg'

image = cv2.imread(image_path)

if image is None:
    print("Image could not be loaded.")
else:
    print("Image loaded successfully.")
```

## 3. Original Image

The original image is displayed before any enhancement is applied. The image has relatively low contrast, making some details less distinct.

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

## 4. Grayscale Conversion

The original image is converted into grayscale so that the image can be represented using pixel intensity values from 0 to 255.

A value close to 0 represents a dark pixel, while a value close to 255 represents a bright pixel.

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8, 6))
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

## 5. Histogram of the Original Image

A histogram represents the distribution of pixel intensity values in an image.

For a low-contrast image, the intensity values are generally concentrated within a limited range. This can make the difference between different areas of the image less noticeable.

```python
plt.figure(figsize=(10, 5))
plt.hist(gray.ravel(), bins=256, range=[0, 256])
plt.title("Histogram of Original Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.show()
```

## 6. Histogram Equalization

Histogram Equalization is used to improve the contrast of the image.

The method changes the distribution of the pixel intensity values so that they are spread over a wider intensity range. This can make details in a low-contrast image more visible.

```python
equalized = cv2.equalizeHist(gray)
```

### Enhanced Image

The result after applying Histogram Equalization is displayed below.

```python
plt.figure(figsize=(8, 6))
plt.imshow(equalized, cmap='gray')
plt.title("Histogram Equalized Image")
plt.axis("off")
plt.show()
```

## 7. Histogram After Equalization

The histogram of the enhanced image is displayed to observe the change in pixel intensity distribution after Histogram Equalization.

```python
plt.figure(figsize=(10, 5))
plt.hist(equalized.ravel(), bins=256, range=[0, 256])
plt.title("Histogram After Histogram Equalization")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.show()
```

## 8. Comparison of Original and Enhanced Images

The original grayscale image is compared with the image after Histogram Equalization to observe the visual improvement in contrast.

```python
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized, cmap='gray')
plt.title("Histogram Equalized Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## 9. Comparison of Histograms

The original and enhanced histograms are compared to see how Histogram Equalization changes the distribution of pixel intensity values.

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.hist(gray.ravel(), bins=256, range=[0, 256])
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")

plt.subplot(1, 2, 2)
plt.hist(equalized.ravel(), bins=256, range=[0, 256])
plt.title("Equalized Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")

plt.tight_layout()
plt.show()
```

## 10. Save the Enhanced Image

The grayscale and enhanced images are saved to the PCD_Assignment02 folder in Google Drive. These files can later be uploaded to the GitHub repository.

```python
output_path = '/content/drive/MyDrive/PCD_Assignment02/'

cv2.imwrite(output_path + 'original_gray.jpg', gray)
cv2.imwrite(output_path + 'histogram_equalized.jpg', equalized)

print("Images saved successfully.")
```

Visually, the enhanced image shows stronger contrast compared to the original grayscale image. Some details that were less visible in the original image become easier to distinguish after enhancement.

The histogram comparison also shows the change in the distribution of pixel intensities before and after Histogram Equalization.
